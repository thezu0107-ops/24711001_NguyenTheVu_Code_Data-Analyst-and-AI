# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python
# TODO
survival_by_class = (
    df.groupby('pclass')['survived']
    .mean()
    .sort_values(ascending=False)
)

print("Tỷ lệ sống sót theo hạng vé:")
print(survival_by_class)

highest_class = survival_by_class.idxmax()
highest_rate = survival_by_class.max()

lowest_class = survival_by_class.idxmin()
lowest_rate = survival_by_class.min()

difference = highest_rate - lowest_rate

print(
    f"\nHạng vé có tỷ lệ sống sót cao nhất: "
    f"{highest_class}"
)

print(
    f"Tỷ lệ sống sót cao nhất: "
    f"{highest_rate * 100:.2f}%"
)

print(
    f"Hạng có tỷ lệ thấp nhất: "
    f"{lowest_class}"
)

print(
    f"Tỷ lệ sống sót thấp nhất: "
    f"{lowest_rate * 100:.2f}%"
)

print(
    f"Chênh lệch: "
    f"{difference * 100:.2f} điểm phần trăm"
)
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
# TODO
survival_by_sex = (
    df.groupby('sex')['survived']
    .mean()
)

print("Tỷ lệ sống sót theo giới tính:")
print(survival_by_sex)

sex_difference = (
    survival_by_sex.max()
    - survival_by_sex.min()
)

print(
    "\nChênh lệch theo giới tính:",
    round(sex_difference * 100, 2),
    "điểm phần trăm"
)
```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
# TODO
survival_by_class = (
    df.groupby('pclass')['survived']
    .mean()
)

class_difference = (
    survival_by_class.max()
    - survival_by_class.min()
)

print(
    "Chênh lệch theo hạng vé:",
    round(class_difference * 100, 2),
    "điểm phần trăm"
)
```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
# TODO
if sex_difference > class_difference:
    print(
        "\n=> Giới tính có ảnh hưởng mạnh hơn "
        "đến tỷ lệ sống sót."
    )
else:
    print(
        "\n=> Hạng vé có ảnh hưởng mạnh hơn "
        "đến tỷ lệ sống sót."
    )
corr_family_survived = (
    df['family_size']
    .corr(df['survived'])
)

print(
    "Hệ số tương quan Family Size - Survived:",
    round(corr_family_survived, 4)
)
```

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

```python
# TODO
survival_by_port = (
    df.groupby('embark_town')['survived']
    .mean()
    .sort_values(ascending=False)
)

print(
    "Tỷ lệ sống sót theo cảng lên tàu:"
)

print(survival_by_port)

best_port = survival_by_port.idxmax()
best_port_rate = survival_by_port.max()

print(
    f"\nCảng có tỷ lệ sống sót cao nhất: "
    f"{best_port}"
)

print(
    f"Tỷ lệ sống sót: "
    f"{best_port_rate * 100:.2f}%"
)
```
