```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path
from scipy import stats

sns.set_style('whitegrid')

csv_path = 'https://raw.githubusercontent.com/databricks/Spark-The-Definitive-Guide/master/data/retail-data/all/online-retail-dataset.csv'

df = pd.read_csv(csv_path, encoding='ISO-8859-1', on_bad_lines='skip')
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print('Loaded from:', csv_path)
df.head()
```

## A.4. Create a new column
Làm sạch dữ liệu (loại Quantity<=0, UnitPrice<=0), tạo cột `Sales` = Quantity * UnitPrice.

```python
# TODO
df_clean = df[
    (df['Quantity'] > 0) &
    (df['UnitPrice'] > 0)
].copy()

df_clean['Sales'] = (
    df_clean['Quantity'] *
    df_clean['UnitPrice']
)

df_clean['InvoiceDate'] = pd.to_datetime(
    df_clean['InvoiceDate']
)

print("Kích thước dữ liệu sau khi làm sạch:",
      df_clean.shape)

print("\n5 dòng đầu:")
print(df_clean.head())
```

# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Quốc gia nào đóng góp doanh thu cao nhất, chiếm bao nhiêu % tổng doanh thu?

```python
# TODO
country_sales = (
    df_clean
    .groupby('Country')['Sales']
    .sum()
    .sort_values(ascending=False)
)

print("Doanh thu theo quốc gia:")
print(country_sales)

top_country = country_sales.idxmax()
top_sales = country_sales.max()

total_sales = df_clean['Sales'].sum()

percentage = top_sales / total_sales * 100

print("\nQuốc gia có doanh thu cao nhất:", top_country)
print(f"Doanh thu: {top_sales:,.2f}")
print(f"Tỷ lệ đóng góp: {percentage:.2f}%")
```

## Câu hỏi 2: Sản phẩm nào bán chạy nhất theo doanh thu?

```python
# TODO
product_sales = (
    df_clean
    .groupby(['StockCode', 'Description'])['Sales']
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 sản phẩm theo doanh thu:")
print(product_sales.head(10))

top_product = product_sales.idxmax()
top_product_sales = product_sales.max()

print("\nSản phẩm có doanh thu cao nhất:")
print("StockCode:", top_product[0])
print("Description:", top_product[1])
print(f"Doanh thu: {top_product_sales:,.2f}")
```

## Câu hỏi 3: Doanh số có tính mùa vụ theo tháng không?

```python
# TODO
df_clean['Month'] = (
    df_clean['InvoiceDate']
    .dt.to_period('M')
)

monthly_sales = (
    df_clean
    .groupby('Month')['Sales']
    .sum()
)

print("Doanh thu theo tháng:")
print(monthly_sales)
```

## Câu hỏi 4: Giá trị đơn hàng trung bình (Average Order Value) khác nhau thế nào giữa các quốc gia?

```python
# TODO
country_aov = (
    df_clean
    .groupby('Country')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Number_of_Orders=('InvoiceNo', 'nunique')
    )
)

country_aov['AOV'] = (
    country_aov['Total_Sales'] /
    country_aov['Number_of_Orders']
)

country_aov = country_aov.sort_values(
    'AOV',
    ascending=False
)

print("AOV theo quốc gia:")
print(country_aov)
```

## Câu hỏi 5: Tỷ lệ giao dịch có dấu hiệu trả hàng/hủy (Quantity âm ở dữ liệu gốc) khác nhau thế nào giữa các quốc gia?

```python
# TODO
df['Is_Return'] = df['Quantity'] < 0

country_transactions = (
    df.groupby('Country')
    .size()
)

country_returns = (
    df[df['Is_Return']]
    .groupby('Country')
    .size()
)

return_rate = (
    country_returns
    .reindex(country_transactions.index, fill_value=0)
    / country_transactions
    * 100
)

return_rate = return_rate.sort_values(
    ascending=False
)

print("Tỷ lệ giao dịch có Quantity âm theo quốc gia:")
print(return_rate)
```
