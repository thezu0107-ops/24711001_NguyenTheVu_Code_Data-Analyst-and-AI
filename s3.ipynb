```python
# chia ra thành 4 nhóm
# 1. đầy đủ count
# 2. Tâm: trung bình, trung vị 
# điểm khác nhau: 
    # - mean: trung bình cộng
    # - median: trung vị
    # - mode: số xuất hiện nhiều nhất

# 3. Biên: min, max, range
# 4. phân tán: std, var, skew, kurt
# - std: độ lệch chuẩn
# - var: phương sai
# - skew: độ lệch
# - kurt: độ nhọn
```

```python
a=pd.Series([1,2,3,4,5])
print(a.var())
print(a.std())
print(a.skew())
print(a.kurt())
```

```python
from scipy import stats

for col in df[['Quantity', 'UnitPrice']]:
    mean_ = df[col].mean()
    median_ = df[col].median()
    skew_ = stats.skew(df[col])
    kurt_ = stats.kurtosis(df[col])  # mặc định đã là Excess Kurtosis (fisher=True)
    print(f"{col}: mean={mean_:.2f}, median={median_:.2f}, skew={skew_:.2f}, kurtosis={kurt_:.2f}")
```

### **Group 2: Data Cleaning (làm sạch dữ liệu)**

Xử lý trùng lặp, missing, kiểu dữ liệu, giao dịch hủy / giá âm, rồi tạo cột `TotalPrice`.

```python
# Làm sạch trên bản sao, giữ nguyên df gốc
df_clean = df.copy()

# 1. Xóa dòng trùng
df_clean = df_clean.drop_duplicates()

# 2. Bỏ dòng thiếu CustomerID hoặc Description
df_clean = df_clean.dropna(subset=["CustomerID", "Description"])

# 3. Đổi kiểu dữ liệu
df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"])
df_clean["CustomerID"] = df_clean["CustomerID"].astype(int)
df_clean["InvoiceNo"] = df_clean["InvoiceNo"].astype(str)
df_clean["Description"] = df_clean["Description"].str.strip()

# 4. Chỉ giữ giao dịch hợp lệ: mua hàng thật (số lượng > 0, giá > 0)
#    InvoiceNo bắt đầu bằng C = đơn hủy
df_clean = df_clean[~df_clean["InvoiceNo"].str.startswith("C")]
df_clean = df_clean[(df_clean["Quantity"] > 0) & (df_clean["UnitPrice"] > 0)]

# 5. Tạo cột doanh thu
df_clean["TotalPrice"] = df_clean["Quantity"] * df_clean["UnitPrice"]

print("Trước khi clean:", df.shape)
print("Sau khi clean :", df_clean.shape)
df_clean.head()
```

```python
# Kiểm tra lại chất lượng sau khi clean
print("Missing còn lại:")
print(df_clean.isnull().sum())

print("\nDòng trùng còn lại:", df_clean.duplicated().sum())
print("Quantity <= 0:", (df_clean["Quantity"] <= 0).sum())
print("UnitPrice <= 0:", (df_clean["UnitPrice"] <= 0).sum())

print("\nTổng doanh thu:", round(df_clean["TotalPrice"].sum(), 2))
print("Số khách hàng:", df_clean["CustomerID"].nunique())
print("Số hóa đơn:", df_clean["InvoiceNo"].nunique())
df_clean.info()
```

```python
# 2. Giá trị thiếu và dòng trùng
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

quality = pd.DataFrame({
    "so_thieu": missing,
    "ty_le_%": missing_pct
})
print("Giá trị thiếu theo cột:")
print(quality[quality["so_thieu"] > 0])

print("\nSố dòng trùng hoàn toàn:", df.duplicated().sum())
print("Tỷ lệ dòng trùng: {:.2f}%".format(df.duplicated().sum() / len(df) * 100))
```
